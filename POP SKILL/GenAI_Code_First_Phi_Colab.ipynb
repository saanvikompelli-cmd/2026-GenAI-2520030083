
##  Load Microsoft Phi Model


```python

from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "microsoft/phi-2"

tokenizer = AutoTokenizer.from_pretrained(
    model_name,
    token=HF_TOKEN
)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    token=HF_TOKEN,
    device_map="auto",
    torch_dtype=torch.float16
)

print(" Model Loaded")
```


##  Helper Function for Text Generation


```python

def run_llm(prompt, max_tokens=200, temp=0.7):

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=max_tokens,
        temperature=temp,
        do_sample=True
    )

    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```


##  Example 1: Simple Question Answering


```python

prompt = "What is machine learning?"

print(run_llm(prompt))
```


##  Example 11: Same Task, Different LLMs - Phi, Qwen


```python

from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

models = {
    "phi": "microsoft/phi-2",
    "qwen": "Qwen/Qwen2-1.5B-Instruct",
    #"mistral": "mistralai/Mistral-7B-Instruct-v0.2"
}

loaded_models = {}

for name, model_id in models.items():

    print(f"Loading {name}...")

    tokenizer = AutoTokenizer.from_pretrained(
        model_id,
        token=HF_TOKEN
    )

    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        token=HF_TOKEN,
        device_map="auto",
        #load_in_8bit=True
    )

    loaded_models[name] = {
        "tokenizer": tokenizer,
        "model": model
    }

print(" All Models Loaded")

def run_model(model_name, prompt, max_tokens=200, temp=0.7):

    tokenizer = loaded_models[model_name]["tokenizer"]
    model = loaded_models[model_name]["model"]

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=max_tokens,
        temperature=temp,
        do_sample=True
    )

    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python

prompt = "Write a Python function to reverse a string."

for m in ["phi","qwen"]:
    print(f"\n--- {m.upper()} ---")
    print(run_model(m, prompt))
```

```python

prompt = '''
If 5 machines make 5 parts in 5 minutes,
how many machines are needed to make 100 parts in 100 minutes?
Solve step by step.
'''

for m in ["phi","qwen"]:
    print(f"\n--- {m.upper()} ---")
    print(run_model(m, prompt))
```
